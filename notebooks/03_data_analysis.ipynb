#  Matplotlib

The foundational plotting library in Python  everything you draw (bar charts, histograms, scatter plots) is built from the same core object model: a figure (the whole canvas) containing one or more axes (the actual plot area with x/y data). It's more manual than Seaborn, but understanding it directly makes Seaborn's shortcuts make sense later

```python
import matplotlib as mlt
import matplotlib.pyplot as plt
```

## 2. Basic line plot

The simplest plot  connects points in order. Good for trends over a continuous variable (like time), though you won't use line plots much in Sprint 02 since your data isn't time-series.

```python
x = [1,2,3,4,5]
y = [10,20,15,25,30]

# print(plt.plot(x, y))
plt.title("Simple Line Plot")
plt.xlabel("x-values")
plt.ylabel("y-values")

plt.savefig("../visuals/line_plot.png")
plt.show()

```

## 3. Bar chart - plt.bar()

Theory: Compares values across categories - this is one of your 4+ required EDA charts. Good for things like "average charges per contract type."

```python
cat = ["Python","java","c++","c#"]
rating =  [0.9,0.6,0.7,0.5]

plt.bar(cat , rating)
plt.title("Market demand tech")
plt.xlabel("Category")
plt.ylabel("Ratings")

plt.savefig("../visuals/bar_chart.png")
plt.show()
```

## 4. Histogram - plt.hist()

Theory: Shows the distribution of a single numeric column - how values are spread out, and where they cluster. Different from a bar chart: bar charts compare separate categories, histograms show the shape of one continuous variable's values (binned into ranges).

```python
data = [22, 25, 22, 30, 35, 28, 40, 22, 30, 45, 25, 33]

plt.hist(data, bins=5)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.savefig("../visuals/Histogram.png")
plt.show()
```

## 5. Scatter plot - plt.scatter()

Theory: Shows the relationship between two numeric variables - each point is one row, positioned by its two values. This is how you visually spot correlation (do higher X values tend to come with higher/lower Y values?).

```python
tenure = [1, 5, 10, 20, 30, 40, 50, 60]
monthly_charges = [20, 35, 45, 60, 55, 70, 65, 80]

plt.scatter(tenure, monthly_charges)
plt.title("Tenure vs Monthly Charges")
plt.xlabel("Tenure (months)")
plt.ylabel("Monthly Charges")
plt.savefig("../visuals/scatter_plot.png")
plt.show()
```

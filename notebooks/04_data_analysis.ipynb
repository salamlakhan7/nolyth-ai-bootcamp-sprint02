# 1. Seaborn 

Built on top of Matplotlib, but designed specifically for statistical/exploratory charts - heatmaps, correlation views, distribution plots - with much less code and better default styling. 

Anything Matplotlib can do manually, Seaborn does with cleaner syntax for the specific chart types that come up constantly in EDA.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

## 2. Distribution plot - sns.histplot()

Same job as Matplotlib's histogram, but with a smoother look and optional density curve built in.

 Use when: same as before - checking the spread/shape of one numeric column.

Real example: distribution of monthly charges.

```python
monthly_charges = [20,35,40,45,50,55,60,65,70,80,85,90,30,42,58,63]

sns.histplot(monthly_charges , bins=8 ,kde=True)# krde purpose to display curve line : bins =no: of values appearnce in img result
plt.title("Distribution of Monthly Charges")
plt.xlabel("Monthly Charges ($)")
plt.savefig("../visuals/seaborn_historgram.png")
plt.show()
```

## 3. Correlation heatmap - sns.heatmap()

This is the single most important Seaborn chart for our churn project. 
It shows the correlation (how strongly two numeric columns move together) between every pair of numeric columns at once, color-coded - dark/light colors for strong positive/negative correlation, neutral color near zero for no relationship.

 Instead of eyeballing dozens of scatter plots one at a time, one heatmap summarizes all relationships together.

Real example: correlation between tenure, monthly charges, and total charges.

How to read it:

data.corr() computes correlation coefficients, ranging from -1 to +1

+1 = perfectly move together (as one goes up, so does the other)

-1 = perfectly opposite (as one goes up, the other goes down)

0 = no relationship at all

annot=True prints the actual number on each cell, not just color

cmap="coolwarm" = warm colors (red) for high positive correlation, cool colors (blue) for negative

```python
data = pd.DataFrame({
    "tenure": [1, 5, 10, 20, 30, 40, 50, 60, 70],
    "MonthlyCharges": [20, 35, 45, 60, 55, 70, 65, 80, 90],
    "TotalCharges": [20, 175, 450, 1200, 1650, 2800, 3250, 4800, 6300]
})

corr = data.corr()
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap ")
plt.savefig("../visuals/seaborn_heatmap.png")
plt.show()
```

## 4. Count plot - sns.countplot()

Shows how many rows fall into each category - literally counting occurrences.

 This is exactly what we'll use for the churn Yes/No breakdown:
 
  how many customers churned vs. stayed, as raw counts instead of a percentage (which is what the pie chart showed earlier).

```python
churn_data = pd.DataFrame({
    "Churn": ["Yes", "No", "No", "No", "Yes", "No", "Yes", "No", "No", "No"]
})

sns.countplot(x="Churn", data=churn_data)
plt.title("Customer Churn Count")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")

plt.savefig("../visuals/seaborn_count.png")
plt.show()
```

## 5. Box plot in Seaborn - sns.boxplot()

Same concept as Matplotlib's box plot (median, quartiles, outliers), but often used to compare distributions across categories side by side -

 which is the more useful version for EDA.

Real example: comparing monthly charges across contract types.

```python
data =pd.DataFrame({
    "Contract": ["Month-to-month","Month-to-month","One year","One year","Two year","Two year","Month-to-month"],
    "MonthlyCharges": [70, 85, 55, 60, 45, 50, 90]
})

sns.boxplot(x="Contract", y="MonthlyCharges", data=data)

plt.title(" Monthly Charges by Contract Type ")
# plt.xlabel(" ")
# plt.ylabel(" ")

plt.savefig("../visuals/seaborn_boxplot.png")
plt.show()
```

#                             Numpy

`The foundation library for numerical computing in Python. Its core feature is the array — a fast, memory-efficient way to store and do math on large sets of numbers all at once, instead of looping through them one by one. Think of it as Python's engine for "do this calculation on a whole table of numbers instantly.`

NumPy handles the raw numbers

```python
import numpy as np
import pandas as pd

print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)
print("Setup working ✅")
```

## 1. Arrays - the core concept

NumPy's whole point is the ndarray (n-dimensional array) - faster and more memory-efficient than Python lists because it stores data in one contiguous block of the same type.

`shape, ndim, dtype, size`

```python
import numpy as np

arr1 = np.array([1,2,3,4])
arr2 = np.array([[1,2],[3,4]]) # D= r*c

print("array 1 :", arr1)
print("array 2 :", arr2)

print("shape of array 1 :", arr1.shape)
print("shape of array 2 :", arr2.shape)

print("N-dimensions of array 1 :", arr1.ndim)
print("N-dimensions of array 2 :", arr2.ndim)

print("Data-Type of array 1 :", arr1.dtype)
print("Data-Type of array 2 :", arr2.dtype)

print("size of array 1 :", arr1.size)
print("size of array 2 :", arr2.size)

```

## 2. Creating arrays (common shortcuts)

```python
a = np.zeros((4,4))
print("creating matrix 4*4 with all 0 elements : \n" , a)

b= np.ones((2,5))
print(" \n creating matrix 2*5 with all 1 elements : \n" , b)

c = np.arange(0,10,2) # (start,stop,step) : working with steppings 
print("\n starting with 0 ,ending at 10(exclude), stepping is 2 : ", c )

d = np.linspace(0,1,5)
print("\n  start from 0  , end at 1 , stepping 5 means there should different/space of 5 : ", d )

id = np.eye(2,2)
print(" \n make an identity matrix : \n ",id)

# Creates a 3x3 identity matrix with 1s on the diagonal
diagonal = np.diag([1, 1, 1])
print("\n set the diagonal elements :\n", diagonal)
```

## 3. Indexing & slicing

```python
ab = np.array([1,3,5,7,9,11,13,15])

print("check element at first(0) index  .....>", ab[0])
print("check element at last(-1) index .....>", ab[-1])
print("check elements from 1st-3rd index .....>", ab[1:3])
print("check all-elements from back-reverse .....>", ab[::-1])

# 2D INDEXING
az = np.array([[2,4,6,8],[10,12,14,16]])

print("\n check element at 1st row with 2nd element : ", az[1,2])
print("\n check each rows first element             : ", az[ :,0]) ## all rows, col 0 → [2,10]
print("\n check each first cols each element        : ", az[0,:]) ## all col, row 0 → [2,4,6,8]

```

## 4. Numerical operations & vectorization

`This is NumPy's superpower — operations apply to the whole array at once, no loops:`

`"why is NumPy faster than a Python loop?" → vectorized operations run in optimized C code under the hood, not the Python interpreter.` 

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

print("\n add both arrays a+b   :" , a+b)
print("\n multiply both arrays  :" ,a*b)
print("\n make a square of a**b :" ,a**b)
print("\n check which elements are greater (boolean answer) :", a>1)
```

## 5. Basic statistics 

```python
sts = np.array([2,4,6,10,12,22,12])

print("Mean(Sum of all elements divided by count) sum/7 :", np.mean(sts))
print("Median:", np.median(sts))
print("Std Dev:", np.std(sts))
print("Min:", np.min(sts))
print("Max:", np.max(sts))
print("Sum:", np.sum(sts))
```

## 6. Reshaping — comes up in ML preprocessing constantly

```python
arr3 = np.arange(12)
reshaped = arr3.reshape(3,4)
print("Reshape gives shape to matrix 3*4 : \n ",reshaped )

flat = reshaped.flatten()
print(" \n flatten reverse the reshaped version :\n",flat)
```

## 7. Broadcasting (important )

NumPy lets you do math between arrays of different shapes without manually looping:

```python
matrix = np.array([[1,2,3],[4,5,6]])
vector = np.array([10, 20, 30])

print(matrix + vector)   # vector added to every row
```

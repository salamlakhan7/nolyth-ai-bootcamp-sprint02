# Pandas 

Built on top of NumPy, but designed for real-world, labeled, messy data — like a spreadsheet living inside Python. Where NumPy gives you raw numerical arrays, Pandas gives you rows and columns with names, and tools built specifically for cleaning and reshaping that kind of data.

The two core structures

1. Series — a single column. One-dimensional, labeled (each value has an index, not just a position). Think of it as one column of a spreadsheet, with row labels attached.

2. DataFrame — a full table. Two-dimensional: rows and columns, where every column is technically a Series, and all columns share the same row index. This is what you'll work with 95% of the time — it is your dataset once loaded.

## 1. Importing Pandas

Theory: Pandas is almost always imported with the alias pd — it's the universal convention, so every tutorial, Stack Overflow answer, and teammate's code will use it too.

```python
import pandas as pd
```

## 2. Series — a single labeled column

A Series is a 1D array with an attached index (labels for each value). Unlike a plain NumPy array, you can access values by label, not just position.

```python
s = pd.Series([10,20,25,30] , index =["a","b","c","d"])
print(f"get series(col) with thier index as mentioned :\n {s}")

print(f"\n using index, call specific series element such as c: {s["c"]}") # access by label → 20
```

## 3.  DataFrame — the full table

A DataFrame is 2D — rows and columns, where each column is a Series. You'll create one manually here just to see the structure, but in real work you almost always load one from a file.

```python
data =  {
    "names"  : ["ab","salam","lakhan","khan"],
    "f_subs" : ["math","science","phyics","chemistry"],
    "stack"  : ["Python","Fastapi","git","DB"]
}
df = pd.DataFrame(data)
print(df)
```

## 4. Reading a CSV — read_csv()

This is how you load a real dataset. Pandas reads the file and automatically infers column names (from the header row) and data types per column — the file becomes a DataFrame in one line.

```python
# python
df = pd.read_csv("data/telco_churn.csv")
```

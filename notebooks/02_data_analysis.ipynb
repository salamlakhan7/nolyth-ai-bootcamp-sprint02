# Pandas 

Built on top of NumPy, but designed for real-world, labeled, messy data — like a spreadsheet living inside Python. Where NumPy gives you raw numerical arrays, Pandas gives you rows and columns with names, and tools built specifically for cleaning and reshaping that kind of data.

The two core structures

1. Series — a single column. One-dimensional, labeled (each value has an index, not just a position). Think of it as one column of a spreadsheet, with row labels attached.

2. DataFrame — a full table. Two-dimensional: rows and columns, where every column is technically a Series, and all columns share the same row index. This is what you'll work with 95% of the time — it is your dataset once loaded.

pd.to_numeric(...) — a Pandas function that tries to convert whatever you give it into numeric values (int or float).

messy["charges"] — the input: this selects the charges column, which is currently text ("100.5", "200", " ", "150.75").

errors="coerce" — this is the important part. It tells Pandas what to do when it hits something it can't convert (like " ", a blank space, which isn't a valid number). There are 3 options for errors:

"raise" (the default) — stop and throw an error the moment it hits something unconvertible
"ignore" — leave the whole column untouched if anything fails (deprecated in newer Pandas, avoid it)
"coerce" — convert everything it can, and turn anything it can't into NaN instead of crashing

So with errors="coerce": "100.5" → 100.5, "200" → 200.0, " " → NaN (since a blank space isn't a number), "150.75" → 150.75.

messy["charges"] = ... — the left side reassigns the result back into the same column, overwriting the old text version with the new numeric version (now containing a real NaN where the bad value was).

Why this matters for your Telco dataset specifically: without errors="coerce", converting TotalCharges would just crash the moment it hit a blank-string row. With coerce, it converts everything valid and cleanly exposes the bad rows as NaN — which .isnull().sum() can then actually catch and count, letting you decide how to handle them (drop the rows, fill with 0, fill with the mean, etc.).

```python
messy = pd.DataFrame({
    "id": [1, 2, 3, 4],
    "charges": ["100.5", "200", " ", "150.75"]
})
print(messy.dtypes)   # charges is 'object' (text), not numeric

messy["charges"] = pd.to_numeric(messy["charges"], errors="coerce")
print(messy)
print("\n messy \n ",messy.dtypes)   # now float64
```

## 11. Filtering - selecting rows by condition

Theory: Same idea as NumPy's boolean indexing, but applied to a full table instead of a flat array. You put a condition inside df[...], and Pandas keeps only the rows where that condition is True.

```python
import numpy as np 
salam = pd.DataFrame({
    "name":  ["Ali", "Sara", "John", "Mira"],
    "city":  ["Karachi", "Lahore", "Karachi", "Lahore"],
    "score": [85, 92, 78, 90]
})

print(salam)  
```

```python
high_score= salam[salam["score"]>=80]
print(high_score)
```

```python
best_city = salam[salam["city"]=="Karachi"]
best_city
```

combining conditions - use & (and) / | (or), each condition in its own parentheses

```python
c_s = salam[(salam["city"]=="Karachi") & (salam["score"]>=80)]
c_s
```

# 12. Grouping - .groupby()

Theory: Splits the data into buckets by a category column, then computes something (mean, sum, count, etc.) separately per bucket. This is how you answer questions like "what's the average score per city?" instead of just one overall average.

```python
avg_by_city = salam.groupby("city")["score"].mean()
print(avg_by_city)

# multiple stats at once
summary = salam.groupby("city")["score"].agg(["mean", "min", "max", "count"])
summary
```

## 13. Sorting - .sort_values()

Theory: Orders rows by a column's values. Useful for spotting top/bottom cases or outliers quickly. ascending=False flips it to highest-first.

```python
print(f"sort the score values :\n {salam.sort_values("score")} ")                 # low to high
print(f"\n sort score values from high to low : \n {salam.sort_values("score", ascending=False)} \n") # high to low
print(f"\n sort by city first ,then score within each city {salam.sort_values(["city", "score"])}")        # sort by city first, then score within each city
```
